```python
import pandas as pd
```

```python
df=pd.read_csv("historical_data (1).csv")
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(
    df['Timestamp IST'],
    format='%d-%m-%Y %H:%M'
).dt.date
```

```python
print(df[['Timestamp IST', 'date']].head())
```

```python
fg = pd.read_csv('fear_greed_index (1).csv')
```

```python
fg['date'] = pd.to_datetime(fg['date']).dt.date
```

```python
print(fg.head())
```

```python
print("Trades range:")
print(df['date'].min(), "to", df['date'].max())

print("\nFear & Greed range:")
print(fg['date'].min(), "to", fg['date'].max())

common = set(df['date']).intersection(set(fg['date']))

print("\nMatching Dates:", len(common))
```

```python
# Cleaned trades file
df.to_csv('trades_cleaned.csv', index=False)

# Cleaned fear & greed file
fg.to_csv('fear_greed_cleaned.csv', index=False)

print("Files Ready ✅")
```
